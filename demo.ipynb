# SpeakerBot Demo & API Interaction
This notebook demonstrates how to interact with your SpeakerBot using standard HTTP requests to the Flask wrapper API!

```python
import requests
import time
import random

# Ensure your Flask app.py is running on port 5000
BASE_URL = 'http://127.0.0.1:5000'

def oled_text(line1, line2):
    res = requests.post(f'{BASE_URL}/api/oled', json={'line1': line1, 'line2': line2})
    return res.json()

def set_servo(angle):
    res = requests.post(f'{BASE_URL}/api/servo', json={'angle': angle})
    return res.json()

def get_mic_amplitude():
    res = requests.get(f'{BASE_URL}/api/mic')
    data = res.json()
    if data.get('esp'):
        return data['esp'].get('amplitude', 0)
    return 0

def stop_music():
    res = requests.post(f'{BASE_URL}/api/stop', json={})
    return res.json()
```

## 1. Test the OLED Display

```python
oled_text("JupyterLite", "Online!")
```

## 2. Dynamic Servo Sweep Animation

```python
# Sweep servo from 0 to 180 degrees
for angle in range(0, 181, 15):
    set_servo(angle)
    time.sleep(0.1)

set_servo(90) # Center the servo back
```

## 3. Mic Event Polling
Here we poll the mic out of the ESP continuously for 20 frames, printing the current noise amplitude.

```python
print("Polling mic...")
for _ in range(20):
    amp = get_mic_amplitude()
    bars = '|' * int(min(50, amp / 50))  # ASCII visualizer
    print(f"{amp:5} {bars}")
    time.sleep(0.1)
```
